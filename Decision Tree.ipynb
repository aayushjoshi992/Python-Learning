A decision tree is a supervised machine learning algorithm used form classification and regression tasks.
It makes decisions by splitting data into smaller subsets based on feature values, creating a tree-like structure.
In a descision Tree, the algorithm decides which feature to split on by measuring how well a feature separates the 
data. The three tyoes if decusion tree are :

    high entropy-mixed classes(impure)
    Low entropy-mostly one class(pure)
