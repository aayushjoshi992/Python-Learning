```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://merolagani.com/LatestMarket.aspx'
response = requests.get(url)

s = BeautifulSoup(response.text, 'html.parser')

details = []
for d in s.find_all("div", class_="table-responsive"):
    for a in d.table.find_all("a", title=True):
        symbol = a['title'].split(' ')[0]
        tr = a.find_parent("tr")         
        ltp = tr.find_all("td")[1].text.strip()   
        details.append({
            'symbol': symbol,
            'ltp':ltp
        })

details
```

```python
df=pd.DataFrame(details)
df
```
