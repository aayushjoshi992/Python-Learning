```python
#Outlier data: very large difference in data
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.DataFrame({
    "Student":["A","B","C","D","E","F","G","H","I","J"],
    "Marks":[5,50,55,60,62,65,68,70,72,150]
})
print(df)
```

```python
plt.boxplot(df['Marks'])
```

```python
Q1=df['Marks'].quantile(0.25)
Q2=df['Marks'].quantile(0.50)
Q3=df['Marks'].quantile(0.75)

IQR=Q3-Q1

print("Q1=",Q1)
print("Q2 median=",Q2)
print("Q3 =",Q3)
print("IQR =", IQR)
```

```python
# find lower and upper bound
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
print("Lower Bound=", lower_bound)
print("Upper Bound=", upper_bound)
```

```python
outliers=df[(df["Marks"]<lower_bound) | (df["Marks"]>upper_bound)]
print("Outliers:")
print(outliers)
```

```python
# Remove Outliers:
clean_df=df[(df["Marks"]>=lower_bound) & 
            (df["Marks"]<=upper_bound)]
print(clean_df)
```

```python
plt.boxplot(clean_df['Marks'])
```
