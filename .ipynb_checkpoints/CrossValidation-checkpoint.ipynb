```python
#Cross-Validation (CV) is a ML technique used to evaluate how well a model
#generalizes to unseen data by repeatedly splitting the available training data into training 
#and validation subsets
```

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_digits
from sklearn.model_selection import (
train_test_split,
StratifiedKFold,
cross_val_score,
GridSearchCV
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
accuracy_score,
classification_report,
confusion_matrix
)
```

```python
digits=load_digits()
digits
```

```python
len(digits)
```

```python
X=digits.data
y=digits.target
print(X.shape,'...',y.shape)
print(np.unique(y))
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,stratify=y,random_state=42)
```

```python
print("\nTraining samples:",X_train.shape[0])
print("Training samples:",X_test.shape[0])
```

```python
sf=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
```

```python
lgr_pipe=Pipeline(
    [
        ('scaler',StandardScaler()),
        ('logr',LogisticRegression(max_iter=2000,random_state=42))
    ]
)
```

```python
# cross validation on training data
```

```python
scores_logr_sf=cross_val_score(lgr_pipe,X_train,y_train,cv=sf,scoring="accuracy")
```

```python
print(f"Fold Score: {scores_logr_sf}")
```

```python
print(f'Mean Value ={scores_logr_sf.mean()}')
```

```python
print("Std CV accuracy:",scores_logr_sf.std())
```

```python
p_grid={
    "logr_C":[0.01,0.1,1,10,100],
    "logr_solver":["lbfgs"],
    "logr_max_iter":[1000,2000]
}
```

```python
grid_search=GridSearchCV(
    estimator= lgr_pipe,
    param_grid= param_grid,
    cv=sf,
    scoring="accuracy",
    n_jobs=1,
    refit=True #
)
```

```python
grid_search.fit(X_train,y_train)
```
