```python
#Exploratory Data Analysis (EDA) is the process of understanding a dataset before
#aplying machine learning or statistical models.
#Two of the most important steps are:
#understanding data types 
#understanding data distribution
# Continuous Data(Decimal value)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.DataFrame({
    "Height":[170.5,168.3,180.9,175.1,169.8],
    "Weight":[65.2,70.1,82.5,80.40,85.0],
    "Temperature":[36.8,37.2,36.5,40.0,42.0]
})
print(df.dtypes)
plt.scatter(df['Height'],df['Weight'])
plt.xlabel("Height")
plt.ylabel("Weight")
plt.show()
```

```python
import seaborn as sns 
df=pd.DataFrame({
    "Department":[
        "IT","HR","IT","Sales","HR",
        "IT","Sales","HR","IT","Sales"
    ]
})
```

```python
sns.countplot(data=df, x="Department")
plt.title("Employee Count by Department")
plt.xlabel("Depratment")
plt.ylabel("Count")
plt.show()
```

```python
#B. Categorical Data::Repesents categories instead of numerical measurements.
```

```python
#1.Nominal Data: male,female,male,red,blue
```

```python
departments=["IT","HR","Sales","Finance"]
employees=[40,20,25,15]
plt.figure(figsize=(6,6,))
plt.pie(employees,labels=departments,autopct="%1.1f%%",startangle=90)
plt.show()
```
