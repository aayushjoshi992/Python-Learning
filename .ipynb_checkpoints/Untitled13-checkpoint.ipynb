```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
```

```python
url='https://www.daraz.com.np/?spm=a2a0e.searchlistcategory.header.dhome.355a5008D0ikWH#?'
response=requests.get(url)
```

```python
s=BeautifulSoup(response.text,'html.parser')
```

```python
products=[]
for p in s.find_all("a",class_="pc-custom-link card-fs-content-body-unit hp-mod-card-hover"):
    title=p.find("p",class_="fs-card-title two-line-clamp").text
    price=p.find("span",class_="price").text
    discount=p.find("span",class_="itemDiscount").text
    products.append(
        {
            'Title':title,
            'Price':price,
            'Discount':discount
        }
    )
products
    
```

```python
df=pd.DataFrame(products)
```
