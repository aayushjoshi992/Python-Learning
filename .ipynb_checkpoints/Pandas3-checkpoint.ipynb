```python
import pandas as pd
```

```python
df1=pd.DataFrame({
    "Name":["Rakesh","Sikha"],
    "Age":[25,30],
    "Id":[10,20]
})
df2=pd.DataFrame({
    "Name":["Anil","Priya"],
    "Age":[22,28],
    "Address":['KTM','Ltp']
})
```

```python
result=pd.concat([df1,df2],ignore_index=True)
result
```
