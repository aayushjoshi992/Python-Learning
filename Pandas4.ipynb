```python
import pandas as pd
```

```python
df=pd.DataFrame({
    "Department":["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Employee":["Rakesh", "Sikha", "Anil", "Priya", "John", "Ram"],
    "Salary":[5000,6000,7000,8000,5500,9000],
    "Age":[25,30,28,35,32,40]
})
```

```python
total_sal_IT=df[df['Department']=='IT']['Salary'].sum()
```

```python
total_sal_HR=df[df['Department']=='HR']['Salary'].sum()
print(f'total HR salary is',total_sal_HR)
```

```python
total_sal_fin=df[df['Department']=='Finance']['Salary'].sum()
```

```python
print(f'total Finance salary is',total_sal_fin)
```

```python
result=df.groupby("Department").agg({
    "Salary":["sum","mean","max"],
    "Age":["min","max"]})
result
```

```python
df2
```
