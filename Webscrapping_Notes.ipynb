Ethics of web scrappping

2.Check robots.txt
Many websites publish a robots.txt file indicating which areas automated systems should avoid (robots.txt is a 
plain text file placed at the root of a website (for exampple, https://rakesh.com/robots.txt) that tells web crawlers
and search engine bots which parts of the site they are allowed or not allowed to access.)

Common directives::
User-agent-specifies which crawler the rules apply to (* means all crawlers).
Disallow- tells crawlers not to visit a path.
Allow- explicitly permits access to a path.
Sitemap- points crawlers to the site's sitemap.

4. Respect privacy:: Do not collect: Personal information, Password, Private user data, Sensetive
information unless you have proper authorization  and a legitimate reason.

6. Follow Copyright and Data Usage Rules::
Even if you can scrape content, you may not have the right to republish or redistribute it. 
Always check: Copyright policies, Licensing terms, Data usage restrictions


## Ethical Web Scraping Checklist::
1. Read terms of Service, 2.Check robots.txt, 3. Use resonable request rates, 
4. Prefer official APIs, 5. Respect privacy and copyrights,6.Store data securely, 7.Avoid disrupting website operations
