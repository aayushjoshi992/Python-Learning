```python
from bs4 import BeautifulSoup
html="""
<html>
<head>
    <title>Learn With Rakesh</title>
</head>
<body>
     <h1>Welcome to Web Scraping</h1>
    <p class="info">Learning BeautifulSoup</p>
    <h3>this is h3 tag in body section</h3>
</body>
<footer><h3>copyright @ rakesh.com</h3></footer>
</html>
"""
print(html)
```

```python
s=BeautifulSoup(html,'html.parser')
```

```python
print(s.title.text)
```

```python
print(s.footer.h3.text)
```

```python
print(s.h1.text)
```

```python
print(s.p.text)
```

```python
allh3=s.find_all('h3')
for t in allh3:
    print(t.text)
```

```python
import requests
from bs4 import BeautifulSoup
url='https://techaxis.com.np/'
response=requests.get(url)
print(response)
```

```python
s=BeautifulSoup(response.text,'html.parser')
print(s.title.text)
```

```python
print(s.body.text)
```

```python
h1=s.find('h2')
print(h1)
```

```python
h2=s.find_all('h2')
if h2:
    print(f'h2 tag found {h2}')
else:
    print('h2 not found')
for h in h2:
    print(h)
```

```python
h3=s.find_all('h3')
for a in h3:
    print(a)
```

```python
para=s.find_all('p')
for p in para:
    print(p)
```
