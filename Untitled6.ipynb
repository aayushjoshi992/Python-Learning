```python
import matplotlib.pyplot as plt
```

```python
revenue=[45000,47000,55000,50000,46000]
year=[2015,2016,2017,2018,2019]
```

```python
plt.plot(year,revenue)
```

```python
revenue=[45000,47000,55000,50000,46000]
expenses=[20000,25000,35000,40000,42000]
bonus=[5000,6000,7000,8000,9000]
year=[2015,2016,2017,2018,2019]
```

```python
plt.plot(year,revenue)
plt.plot(year,expenses)
plt.plot(year,bonus)
plt.style.use('default')
plt.xticks(year)
plt.show()
```

```python
revenue=[45000,47000,55000,50000,46000]
expenses=[20000,25000,35000,40000,42000]
bonus=[5000,6000,7000,8000,9000]
year=[2015,2016,2017,2018,2019]

plt.plot(year,revenue,color='red')
plt.plot(year,expenses,color='green')
plt.plot(year,bonus,color='silver')
plt.title('Revenue vs Expenses over a year')
plt.xlabel('Year')
plt.ylabel('Revenue vs Expense')
plt.style.use('default')
plt.xticks(year)
plt.show()
```

```python
revenue=[45000,47000,55000,50000,46000]
expenses=[20000,25000,35000,40000,42000]
bonus=[5000,6000,7000,8000,9000]
year=[2015,2016,2017,2018,2019]

plt.plot(year,revenue,color='red',linestyle='solid',linewidth=2)
plt.plot(year,expenses,color='green',linestyle='dashed',linewidth=3)
plt.plot(year,bonus,color='silver',linestyle='dotted',linewidth=4)
plt.title('Revenue vs Expenses over a year')
plt.xlabel('Year')
plt.ylabel('Revenue vs Expense')
plt.style.use('default')
plt.xticks(year)
plt.show()
```

```python
revenue=[45000,47000,55000,50000,46000]
expenses=[20000,25000,35000,40000,42000]
bonus=[5000,6000,7000,8000,9000]
year=[2015,2016,2017,2018,2019]

plt.plot(year,revenue,color='red',linestyle='solid',linewidth=2,marker='D',markersize=8)
plt.plot(year,expenses,color='green',linestyle='dashed',linewidth=3,marker='o',markersize=8)
plt.plot(year,bonus,color='silver',linestyle='dotted',linewidth=4,marker='s',markersize=7)
plt.title('Revenue vs Expenses over a year')
plt.xlabel('Year')
plt.ylabel('Revenue vs Expense')
plt.style.use('default')
plt.xticks(year)
plt.show()
```

```python
revenue=[45000,47000,55000,50000,46000]
expenses=[20000,25000,35000,40000,42000]
bonus=[5000,6000,7000,8000,9000]
year=[2015,2016,2017,2018,2019]

plt.plot(year,revenue,color='red',linestyle='solid',linewidth=2,marker='D',markersize=8,label='Revenue')
plt.plot(year,expenses,color='green',linestyle='dashed',linewidth=3,marker='o',markersize=8,label='Expense')
plt.plot(year,bonus,color='silver',linestyle='dotted',linewidth=4,marker='s',markersize=7,label='Bonus')
plt.title('Revenue vs Expenses over a year')
plt.xlabel('Year')
plt.ylabel('Revenue vs Expense')
plt.style.use('default')
plt.xticks(year)
plt.legend(loc='best')
plt.grid(True, linestyle='--', alpha=0.9)
plt.show()
```
