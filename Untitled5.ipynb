```python
import pandas as pd
```

```python
df1=pd.DataFrame({
    "Name":["Rakesh","Sikha"],
    "Age":[25,30],
    "Id":[10,20]
})
df2=pd.DataFrame({
    "Name":["Anil","Priya"],
    "Age":[22,28],
    "Address":['KTM','Ltp']
})
```

```python
result=pd.concat([df1,df2],ignore_index=True)
result
```

```python
df3=pd.DataFrame({
    "Salary":[5000,7000]
})
res=pd.concat([df1,df3],axis=1)
res
```

```python
employees=pd.DataFrame({
    "EmpID":[1,2,3],
    "Name":["Rakesh","Sikha","Anil"],
})
salaries=pd.DataFrame({
    "EmpID":[1,2,4],
    "Salary":[5000,7000,6000]
})

```

```python
merged=pd.merge(employees,salaries)
merged
```

```python
merged=pd.merge(employees,salaries, on='EmpID')
merged
```

```python
result=pd.merge(employees,salaries,on='EmpID', how='left')
result
```

```python
result=pd.merge(employees,salaries,on='EmpID', how='right')
result
```

```python
result=pd.merge(employees,salaries,on='EmpID', how='outer')
result
```

```python
left=pd.DataFrame({
    "Name":["Rakesh","Sikha","Anil"],
},index=[1,2,3])
right=pd.DataFrame({
    "Salary":[5000,7000,6000]
},index=[1,2,4])
```

```python
res=left.join(right)
```

```python
res
```

```python
inner=left.join(right, how="inner")#only columns
```
