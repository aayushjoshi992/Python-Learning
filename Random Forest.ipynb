One line summary
Random Forest builds many decision trees using different samples.


Hyperparamenter
These are three important Random forest hyperpramaeter that control how the trees are built

1. min_samples_split=2
**Meaning**
The minimum number of samples required at a node before the node can be split into two child nodes.
Default=2
The algorithm check every node.
If the node has at least 2 samples, it is allowed to split.

2.min_samples_leaf=1
The minimum number of samples that must remain in every leaft node after a split.


3.max_feature="sqrt"
Only 4 randomly chosen features are evaluated at each split.
