```python
import matplotlib.pyplot as plt
```

```python
#A pie chart: is used to show part-to-whole relationships, meaning how a total is 
#divided into different categories #Each slice represent a percentage of the whole
```

```python
subject=["Java","Python","C","JavaScript"]
marks=[50,35,30,18]
plt.figure(figsize=(6,4))
plt.pie(marks,labels=subject, autopct="%1.1f%%",startangle=90)
plt.title('Percentage programming language distribution')
plt.axis('equal')
plt.show()
```

```python
subjects=["Java","Python","C","JavaScript","PHP"]
marks=[50,35,30,18,60]
#Highlight Python slice
explodeval=[0.1,0,0,0.1,0]
plt.figure(figsize=(6,4))
plt.pie(marks,labels=subjects,autopct="%1.1f%%", explode=explodeval,shadow=True,startangle=90)
plt.title("Percentage Programming Language Distribution")
plt.axis("equal")
plt.savefig('piemarks.png')
plt.show()
```

```python
tasks=['Completed','In Progress','Not Started','On Hold']
task_counts=[45,30,15,10]

#2. Customize the visuals presentation
colors=['#4CAF50', '#FFC107', '#F44336', '#12544F']
explode=(0.1,0,0,0) #Explode/ Separate the 1st slice('Completed')

#3. Create the pie chart
plt.figure(figsize=(8,6))
plt.pie(task_counts,
        labels=tasks,
        colors=colors,
        explode=explode,
        autopct='%1.1f%%',   #Formats and displays the numerical percentages
        startangle=140,      # Rotates the start position of the pie chart
        shadow=True,         # Adds a subtle 3D drop shadow effect
        textprops={'fontsize':12}  #Sets font configuration for the labels
)
plt.title('Project Task Statis Distribution', fontsize=16,fontweight='bold')
plt.tight_layout()
plt.show()
```
