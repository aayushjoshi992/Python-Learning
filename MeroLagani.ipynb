```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://merolagani.com/LatestMarket.aspx'
response = requests.get(url)

s = BeautifulSoup(response.text, 'html.parser')

details = []
for d in s.find_all("div", id="ctl00_ContentPlaceHolder1_LiveTrading"):
    for a in d.table.find_all("a", title=True):
        symbol = a['title'].split(' ')[0]
        tr = a.find_parent("tr")         
        ltp = tr.find_all("td")[1].text.strip()
        change=tr.find_all("td")[2].text.strip()
        op=tr.find_all("td")[3].text
        high=tr.find_all("td")[4].text.strip()
        low=tr.find_all("td")[5].text.strip()
        Qty=tr.find_all("td")[6].text.strip()
        pClose=tr.find_all("td")[7].text.strip()
        diff=tr.find_all("td")[8].text.strip()
       
        details.append({
            'symbol': symbol,
            'ltp':ltp,
            'change':change,
            'open':op,
            'high':high,
            'low':low,
            'Qty':Qty,
            'pClose':pClose,
            'Diff':diff
            
        })

details
```

```python
df=pd.DataFrame(details)
df
```
