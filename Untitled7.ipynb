```python
#Scatter Plot: A scatter plot is a type of datat graph that uses data to show the relationship between two numeric variable
```

```python
import matplotlib.pyplot as plt
```

```python
#Sample data
study_hours=[1,2,3,4,5,6,7]
exam_scores=[55,60,65,72,78,85,90]
```

```python
plt.plot(study_hours,exam_scores)
```

```python
plt.figure(figsize=(6,4))
plt.scatter(study_hours,exam_scores)
plt.title("Study Hours vs Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.DataFrame({
    "Age":[20,25,30,35,40],
    "Salary":[30000,40000,50000,650000,80000]
})
plt.figure(figsize=(6,4))
plt.scatter(df["Age"],df["Salary"])
plt.plot(df['Age'],df['Salary'])
plt.show()

```
