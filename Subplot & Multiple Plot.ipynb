```python
# Subplots and multiple plots are used to display more than one graph in a figure.
# 1. Subplot: A subplot divides a single figure into multiple smaller plots arranged
# in rows and columns. 
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y1=[2,4,6,8,10]
y2=[1,4,9,16,25]
```

```python
plt.figure(figsize=(6,4))

#First subplot: subplot(nrows, ncols, index)
plt.subplot(1,1,1)
plt.plot(x,y1,marker='o')
plt.title("Linear Growth")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

#Second plot
plt.subplot(2,1,2)
plt.plot(x,y2,marker='s')
plt.title("Quadratic Growth")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

plt.tight_layout(pad=3)
plt.show()
```

```python
x=[1,2,3,4,5]
y1=[2,4,6,8,10]
y2=[1,4,9,16,25]
y3=[5,10,15,20,25]
fig,ax=plt.subplots(1,3,figsize=(10,4))#design=fig,ax= array of immage kati ota image banna aauncha =3
fig.set_facecolor("orange")
ax[0].plot(x,y1,marker='o')
ax[0].set_title('linear')
ax[0].grid(True)

ax[1].plot(x,y2,marker='s')
ax[1].set_title('Quadratic')
ax[1].grid(True)

ax[2].plot(x,y3,marker='d')
ax[2].set_title('Quadratic')
ax[2].grid(True)
plt.tight_layout()
plt.savefig('subplots.png')
plt.show()

```
